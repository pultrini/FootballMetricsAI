```python
shots[
    [
        "player",
        "location",
        "shot_statsbomb_xg",
        "shot_outcome",
        "goal",
    ]
].head(20)
```

```python
shots["x"] = shots["location"].apply(
    lambda location: location[0]
)

shots["y"] = shots["location"].apply(
    lambda location: location[1]
)

plt.figure(figsize=(12, 8))

plt.scatter(
    shots["x"],
    shots["y"],
    s=shots["shot_statsbomb_xg"] * 500 + 10,
    alpha=0.6,
)

plt.xlim(0, 120)
plt.ylim(0, 80)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Finalizações — tamanho = StatsBomb xG")

plt.show()
```

```python
goals = shots[shots["goal"] == 1]
misses = shots[shots["goal"] == 0]

plt.figure(figsize=(12, 8))

plt.scatter(
    misses["x"],
    misses["y"],
    s=misses["shot_statsbomb_xg"] * 400 + 10,
    alpha=0.4,
    label="Não gol",
)

plt.scatter(
    goals["x"],
    goals["y"],
    s=goals["shot_statsbomb_xg"] * 400 + 10,
    alpha=0.9,
    label="Gol",
)

plt.xlim(60, 120)
plt.ylim(0, 80)

plt.xlabel("X")
plt.ylabel("Y")

plt.title("Mapa de finalizações")
plt.legend()

plt.show()
```

```python
from statsbombpy import sb

events = sb.competition_events(
    country="International",
    division="FIFA World Cup",
    season="2022",
    gender="male",
)

shots = events[
    events["type"] == "Shot"
].copy()

shots["goal"] = (
    shots["shot_outcome"] == "Goal"
).astype(int)

shots.shape
```

```python
shots.groupby("goal")["shot_statsbomb_xg"].describe()
```

```python
shots["goal"].value_counts()
```

```python
import pandas as pd

shots["xg_bin"] = pd.cut(
    shots["shot_statsbomb_xg"],
    bins=[
        0.0,
        0.05,
        0.10,
        0.20,
        0.30,
        0.40,
        0.50,
        0.75,
        1.0,
    ],
)
```

```python
calibration = (
    shots
    .groupby("xg_bin", observed=True)
    .agg(
        shots=("goal", "size"),
        goals=("goal", "sum"),
        actual_goal_rate=("goal", "mean"),
        mean_xg=("shot_statsbomb_xg", "mean"),
    )
)

calibration
```

```python
import cv2

saida = "/Users/davipultrini/Documents/studies/python_projects/StudingDataScience/footballAI/data/0bfacc_0_12fps.mp4"

cap = cv2.VideoCapture("/Users/davipultrini/Documents/studies/python_projects/StudingDataScience/footballAI/data/0bfacc_0.mp4")
fps = 12

if not cap.isOpened():
    raise RuntimeError("não foi possível abrir o vídeo")

fps_original = cap.get(cv2.CAP_PROP_FPS)
largura = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
altura = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*"mp4v")

out = cv2.VideoWriter(
    saida,
    fourcc,
    fps,
    (largura, altura)
)

contador= 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    if contador %5 == 0:
        out.write(frame)
    
    contador += 1

cap.release()
out.release()

print("video salvo")
```
